# Section 5: Data Science
The question from section 4 that I will examine is "Can we build a multivariate regression model to predict wine quality based on chemical properties?". The reason why I want to answer this question specifically is because quality is subjective, so I want to explore what makes wine high quality generally. I outlined a potential solution in the question itself, using a regression model, but the main thing would be to decide: what variables should I use? We'll probably want to use variables that are better predictors of quality rather than throwing everything together, since it wouldn't make sense to use bad predictors to predict quality.

Based on my exploratory data analysis, I'll include the following variables:
- Alcohol (r=0.48) - strongest positive predictor
- Volatile acidity (r=-0.39) - strongest negative predictor  
- Sulphates (r=0.25) - moderate positive predictor

And then I'll evaluate model performance with R2 score.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
df.columns.tolist()
```

```python
# make the model
ind = ["alcohol","volatile acidity","sulphates"]
dep = "quality"
model = LinearRegression()
model.fit(df[ind], df[dep])
df["predicted quality"] = model.predict(df[ind])
df
```

```python
# evaluate the model
from sklearn.metrics import r2_score
r2 = r2_score(df[dep], df["predicted quality"])
r2
```

I created a model that achieves an R2 of 0.336, meaning that the model explains approximately 34% of the variance in wine quality. This is indicative of a moderate prediction strength, but it makes sense because wine quality is mostly based on human tastes which is subjective. Therefore, chemical composition provides good but also incomplete insight on whether wine is good or bad.

# Section 6: A Different Data Visualization

```python
df['residual'] = df['quality'] - df['predicted quality']
df['residual'].plot.hist(bins=30, edgecolor='black', alpha=0.7, figsize=(8, 6),xlabel="residual (predicted - actual)",title="distribution of model prediction errors for wine quality")
```

Interpretation: the histogram shows that the prediction errors are approximately normally distributed around 0 with most errors being between -1 and 1. This shows that the model can make reasonable predictions without much bias
